```python
import warnings
warnings.filterwarnings("ignore")

import os
import json
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              roc_auc_score, confusion_matrix, classification_report,
                              roc_curve)
from sklearn.decomposition import PCA

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", DEVICE)

sns.set_theme(style="whitegrid")
```

```python
DATA_PATH = "Dataset.csv"

import os
import pandas as pd

# Assuming Dataset.csv is now manually uploaded or present in the environment.
# If you encounter FileNotFoundError, please ensure the file is uploaded.

try:
    df_raw = pd.read_csv(DATA_PATH)
    if df_raw.empty:
        raise ValueError("Uploaded CSV file is empty or contains no data rows after reading.")
    print(df_raw.shape)
    display(df_raw.head())
except FileNotFoundError:
    raise RuntimeError(f"'{DATA_PATH}' not found. Please upload the 'Dataset.csv' file to your Colab environment or ensure it is in the correct path.")
except pd.errors.EmptyDataError:
    raise RuntimeError("Uploaded file is empty or only contains headers, could not read CSV.")
except pd.errors.ParserError as e:
    raise RuntimeError(f"Error parsing uploaded CSV file: {e}")
except Exception as e:
    print(f"An unexpected error occurred while reading the CSV: {e}")
    raise
```

```python
# Courses treated as "gateway" / foundational per the proposal (Section 5.1): first-year
# math, intro programming, discrete structures and data-structures-style courses whose
# early failure is hypothesized to predict later dropout.
GATEWAY_COURSES = ["MAT 101", "MAT 102", "CSE 100", "CSE 101", "CSE 102",
                    "CSE 111", "CSE 112", "CSE 215", "CSE 221"]

STATIC_COLS = [
    "Student_ID", "Admission_Year", "Age", "Gender", "Scholarship_Status",
    "Current_Semester", "Current_CGPA", "Courses_Per_Semester", "Failed_Courses_Count",
    "Credit_Completed", "Credit_Required", "Academic_Probation", "Course_Withdrawal_Count",
    "Exam_Miss_Count", "LMS_Login_Frequency_Weekly", "LMS_Total_Time_Hours",
    "Days_Since_Last_Login", "Assignment_Submission_Count", "OnTime_Submission_Ratio",
    "Quiz_Attempt_Count", "Quiz_Average_Score", "Forum_View_Count", "Missed_Deadlines_Count",
    "Attendance_Pct", "Lab_Attendance_Pct", "Attendance_Trend", "Fee_Payment_Status",
    "Fee_Payment_Delay_Days", "Semesters_On_Probation", "Extracurricular_Participation",
    "Dropout_Risk_Level", "Dropout_Status", "Semester_GPA", "Courses_Failed_This_Semester",
    "Consecutive_Absence_Streak", "Gap_Semester_Flag", "Credit_Load_Enrolled_This_Semester",
    "Resource_Download_Count", "Program_Track",
]

def aggregate_to_student_level(df):
    # Check if the input DataFrame is valid for aggregation
    required_cols_for_agg = [
        "Student_ID", "Course_Name", "Course_Grade", "Course_Grade_Point",
        "Retake_Flag", "Forum_Post_Text"
    ] + STATIC_COLS # STATIC_COLS contains Student_ID

    missing_cols = [col for col in required_cols_for_agg if col not in df.columns]

    if df.empty or missing_cols:
        print(f"Warning: Input DataFrame for aggregation is empty or missing required columns: {missing_cols}")
        print("Returning an empty student_df DataFrame to prevent further errors.")
        # Define all columns that `student_df` is expected to have
        expected_student_df_cols = list(STATIC_COLS) + [
            "Num_Courses_Taken", "Num_F_Grades", "Avg_Grade_Point", "Retake_Count",
            "Gateway_Course_Failure_Flag", "Forum_Text_Combined"
        ]
        return pd.DataFrame(columns=expected_student_df_cols)

    static = df.groupby("Student_ID")[STATIC_COLS].first().reset_index(drop=True)

    course_agg = df.groupby("Student_ID").agg(
        Num_Courses_Taken=("Course_Name", "count"),
        Num_F_Grades=("Course_Grade", lambda x: (x == "F").sum()),
        Avg_Grade_Point=("Course_Grade_Point", "mean"),
        Retake_Count=("Retake_Flag", lambda x: (x == "Yes").sum()),
    ).reset_index()

    # A student is flagged if ANY of their course rows is both a gateway course and
    # graded F.
    gw = df.copy()
    gw["is_gateway_fail"] = gw["Course_Name"].isin(GATEWAY_COURSES) & (gw["Course_Grade"] == "F")
    gw_flag = gw.groupby("Student_ID")["is_gateway_fail"].max().astype(int).reset_index()
    gw_flag = gw_flag.rename(columns={"is_gateway_fail": "Gateway_Course_Failure_Flag"})

    course_agg = course_agg.merge(gw_flag, on="Student_ID")

    text_agg = df.groupby("Student_ID")["Forum_Post_Text"].apply(
        lambda posts: " . ".join(str(p) for p in posts if isinstance(p, str) and p.strip())
    ).reset_index().rename(columns={"Forum_Post_Text": "Forum_Text_Combined"})

    student_df = static.merge(course_agg, on="Student_ID").merge(text_agg, on="Student_ID")
    return student_df

student_df = aggregate_to_student_level(df_raw)
print("Aggregated shape:", student_df.shape)
student_df.head()
```

```python
student_df["CGPA_Delta"] = student_df["Current_CGPA"] - student_df["Semester_GPA"]
student_df["Attendance_Drop_Indicator"] = (student_df["Attendance_Trend"] == "Decreasing").astype(int)
student_df["Cumulative_Failure_Load"] = student_df["Failed_Courses_Count"] + student_df["Num_F_Grades"]
student_df["Credit_Load_Ratio"] = (
    student_df["Credit_Load_Enrolled_This_Semester"] / student_df["Credit_Required"].replace(0, np.nan)
).fillna(0)
student_df["Submission_Lateness_Ratio"] = 1 - student_df["OnTime_Submission_Ratio"]

# Semester bucket used later for the "earliest reliable prediction point" analysis
student_df["Semester_Bucket"] = pd.cut(
    student_df["Current_Semester"], bins=[0, 3, 5, 8],
    labels=["Early (1-3)", "Mid (4-5)", "Late (6-8)"]
)

student_df["Target"] = (student_df["Dropout_Status"] == "Yes").astype(int)

print(student_df["Target"].value_counts(normalize=True))
student_df[["CGPA_Delta", "Attendance_Drop_Indicator", "Cumulative_Failure_Load",
            "Credit_Load_Ratio", "Submission_Lateness_Ratio", "Gateway_Course_Failure_Flag"]].describe()
```

```python
from transformers import AutoTokenizer, AutoModel, AutoModelForSequenceClassification

if student_df.empty:
    print("Warning: student_df is empty. Skipping text feature processing.")
    # Initialize columns that would normally be created so downstream cells don't error
    student_df["Forum_Sentiment_Score"] = np.nan
    for i in range(16):
        student_df[f"Text_Embed_{i}"] = np.nan
    # Exit early from this cell
    raise ValueError("student_df is empty. Please ensure Dataset.csv was loaded correctly.")

TEXT_MODEL_NAME = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(TEXT_MODEL_NAME)
sentiment_model = AutoModelForSequenceClassification.from_pretrained(TEXT_MODEL_NAME).to(DEVICE).eval()
embed_model = AutoModel.from_pretrained("distilbert-base-uncased").to(DEVICE).eval()

@torch.no_grad()
def batch_text_features(texts, batch_size=64, max_len=128):
    sentiment_scores = []
    embeddings = []
    for i in range(0, len(texts), batch_size):
        batch = [t if isinstance(t, str) and t.strip() else "no comments available" for t in texts[i:i+batch_size]]
        enc = tokenizer(batch, padding=True, truncation=True, max_length=max_len, return_tensors="pt").to(DEVICE)

        # sentiment
        logits = sentiment_model(**enc).logits
        probs = torch.softmax(logits, dim=-1)  # [neg, pos] for this checkpoint
        score = (probs[:, 1] - probs[:, 0]).cpu().numpy()
        sentiment_scores.append(score)

        # mean-pooled embedding from the base DistilBERT encoder
        out = embed_model(input_ids=enc["input_ids"], attention_mask=enc["attention_mask"])
        mask = enc["attention_mask"].unsqueeze(-1).float()
        pooled = (out.last_hidden_state * mask).sum(1) / mask.sum(1).clamp(min=1e-9)
        embeddings.append(pooled.cpu().numpy())

        if i % (batch_size * 20) == 0:
            print(f"  processed {i}/{len(texts)} students' text")

    return np.concatenate(sentiment_scores), np.concatenate(embeddings, axis=0)

texts = student_df["Forum_Text_Combined"].tolist()
sent_scores, text_embeddings = batch_text_features(texts)

student_df["Forum_Sentiment_Score"] = sent_scores

pca = PCA(n_components=16, random_state=SEED)
text_embed_reduced = pca.fit_transform(text_embeddings)
print("Explained variance (16 PCA components):", pca.explained_variance_ratio_.sum().round(3))

text_embed_cols = [f"Text_Embed_{i}" for i in range(16)]
student_df[text_embed_cols] = text_embed_reduced
student_df[["Forum_Sentiment_Score"] + text_embed_cols[:3]].describe()
```

```python
CATEGORICAL_COLS = [
    "Gender", "Scholarship_Status", "Academic_Probation", "Attendance_Trend",
    "Fee_Payment_Status", "Extracurricular_Participation", "Gap_Semester_Flag",
    "Dropout_Risk_Level", "Program_Track",
]

NUMERIC_COLS = [
    "Admission_Year", "Age", "Current_Semester", "Current_CGPA", "Courses_Per_Semester",
    "Failed_Courses_Count", "Credit_Completed", "Credit_Required", "Course_Withdrawal_Count",
    "Exam_Miss_Count", "LMS_Login_Frequency_Weekly", "LMS_Total_Time_Hours",
    "Days_Since_Last_Login", "Assignment_Submission_Count", "OnTime_Submission_Ratio",
    "Quiz_Attempt_Count", "Quiz_Average_Score", "Forum_View_Count", "Missed_Deadlines_Count",
    "Attendance_Pct", "Lab_Attendance_Pct", "Fee_Payment_Delay_Days", "Semesters_On_Probation",
    "Semester_GPA", "Courses_Failed_This_Semester", "Consecutive_Absence_Streak",
    "Credit_Load_Enrolled_This_Semester", "Resource_Download_Count", "Num_Courses_Taken",
    "Num_F_Grades", "Avg_Grade_Point", "Retake_Count", "Gateway_Course_Failure_Flag",
    "CGPA_Delta", "Attendance_Drop_Indicator", "Cumulative_Failure_Load",
    "Credit_Load_Ratio", "Submission_Lateness_Ratio", "Forum_Sentiment_Score",
] + text_embed_cols

label_encoders = {}
for col in CATEGORICAL_COLS:
    le = LabelEncoder()
    student_df[col] = le.fit_transform(student_df[col].astype(str))
    label_encoders[col] = le

CAT_CARDINALITIES = [student_df[c].nunique() for c in CATEGORICAL_COLS]
print(dict(zip(CATEGORICAL_COLS, CAT_CARDINALITIES)))

X = student_df[CATEGORICAL_COLS + NUMERIC_COLS].copy()
y = student_df["Target"].values
semester_bucket = student_df["Semester_Bucket"].values  # kept aside for Section 9

X_train, X_test, y_train, y_test, bucket_train, bucket_test = train_test_split(
    X, y, semester_bucket, test_size=0.30, random_state=SEED, stratify=y
)

print("Train class balance:", np.bincount(y_train) / len(y_train))
print("Test class balance:", np.bincount(y_test) / len(y_test))
```

```python
from imblearn.over_sampling import SMOTENC

cat_indices = list(range(len(CATEGORICAL_COLS)))  # categorical cols are first N columns of X

smote = SMOTENC(categorical_features=cat_indices, random_state=SEED)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
print("Resampled train class balance:", np.bincount(y_train_res) / len(y_train_res))

# Standardize numeric columns AFTER resampling, fit on training data only, applied to
# both the resampled training set and the untouched test set.
scaler = StandardScaler()
X_train_res[NUMERIC_COLS] = scaler.fit_transform(X_train_res[NUMERIC_COLS])
X_test_scaled = X_test.copy()
X_test_scaled[NUMERIC_COLS] = scaler.transform(X_test_scaled[NUMERIC_COLS])
```

```python
class TabularDataset(Dataset):
    def __init__(self, X_cat, X_num, y=None):
        self.X_cat = torch.tensor(X_cat, dtype=torch.long)
        self.X_num = torch.tensor(X_num, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32) if y is not None else None

    def __len__(self):
        return len(self.X_cat)

    def __getitem__(self, idx):
        if self.y is not None:
            return self.X_cat[idx], self.X_num[idx], self.y[idx]
        return self.X_cat[idx], self.X_num[idx]


class FTTransformer(nn.Module):
    def __init__(self, cat_cardinalities, n_numeric, d_token=32, n_layers=3, n_heads=4,
                 dropout=0.15):
        super().__init__()
        self.n_cat = len(cat_cardinalities)
        self.n_num = n_numeric
        self.d_token = d_token

        # one embedding table per categorical feature
        self.cat_embeddings = nn.ModuleList([
            nn.Embedding(card, d_token) for card in cat_cardinalities
        ])
        # one linear "tokenizer" per numeric feature: scalar -> d_token vector
        self.num_weight = nn.Parameter(torch.randn(n_numeric, d_token) * 0.02)
        self.num_bias = nn.Parameter(torch.zeros(n_numeric, d_token))

        self.cls_token = nn.Parameter(torch.randn(1, 1, d_token) * 0.02)

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_token, nhead=n_heads, dim_feedforward=d_token * 4,
            dropout=dropout, activation="gelu", batch_first=True,
        )
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.norm = nn.LayerNorm(d_token)
        self.head = nn.Sequential(
            nn.Linear(d_token, d_token), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(d_token, 1),
        )

    def forward(self, x_cat, x_num):
        b = x_cat.size(0)
        cat_tokens = torch.stack(
            [emb(x_cat[:, i]) for i, emb in enumerate(self.cat_embeddings)], dim=1
        )  # (batch, n_cat, d_token)
        num_tokens = x_num.unsqueeze(-1) * self.num_weight + self.num_bias  # (batch, n_num, d_token)

        cls = self.cls_token.expand(b, -1, -1)
        tokens = torch.cat([cls, cat_tokens, num_tokens], dim=1)

        encoded = self.encoder(tokens)
        cls_out = self.norm(encoded[:, 0])
        logits = self.head(cls_out).squeeze(-1)
        return logits
```

```python
n_cat = len(CATEGORICAL_COLS)
n_num = len(NUMERIC_COLS)

X_train_cat = X_train_res[CATEGORICAL_COLS].values.astype(np.int64)
X_train_num = X_train_res[NUMERIC_COLS].values.astype(np.float32)
X_test_cat = X_test_scaled[CATEGORICAL_COLS].values.astype(np.int64)
X_test_num = X_test_scaled[NUMERIC_COLS].values.astype(np.float32)

train_ds = TabularDataset(X_train_cat, X_train_num, y_train_res)
test_ds = TabularDataset(X_test_cat, X_test_num, y_test)

train_loader = DataLoader(train_ds, batch_size=256, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=512, shuffle=False)

model = FTTransformer(cat_cardinalities=CAT_CARDINALITIES, n_numeric=n_num).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=20)
criterion = nn.BCEWithLogitsLoss()

print(model)
print("Trainable parameters:", sum(p.numel() for p in model.parameters() if p.requires_grad))
```

```python
N_EPOCHS = 20
history = {"train_loss": [], "val_auc": []}

def evaluate(loader):
    model.eval()
    all_probs, all_labels = [], []
    with torch.no_grad():
        for x_cat, x_num, yb in loader:
            x_cat, x_num = x_cat.to(DEVICE), x_num.to(DEVICE)
            logits = model(x_cat, x_num)
            probs = torch.sigmoid(logits).cpu().numpy()
            all_probs.append(probs)
            all_labels.append(yb.numpy())
    return np.concatenate(all_probs), np.concatenate(all_labels)

for epoch in range(1, N_EPOCHS + 1):
    model.train()
    running_loss = 0.0
    for x_cat, x_num, yb in train_loader:
        x_cat, x_num, yb = x_cat.to(DEVICE), x_num.to(DEVICE), yb.to(DEVICE)
        optimizer.zero_grad()
        logits = model(x_cat, x_num)
        loss = criterion(logits, yb)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        running_loss += loss.item() * x_cat.size(0)
    scheduler.step()

    train_loss = running_loss / len(train_ds)
    val_probs, val_labels = evaluate(test_loader)
    val_auc = roc_auc_score(val_labels, val_probs)
    history["train_loss"].append(train_loss)
    history["val_auc"].append(val_auc)
    print(f"Epoch {epoch:2d}/{N_EPOCHS} | train_loss={train_loss:.4f} | test_AUC={val_auc:.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history["train_loss"]); axes[0].set_title("Training loss"); axes[0].set_xlabel("epoch")
axes[1].plot(history["val_auc"], color="darkorange"); axes[1].set_title("Test AUC"); axes[1].set_xlabel("epoch")
plt.tight_layout(); plt.show()
```

```python
test_probs, test_labels = evaluate(test_loader)
test_preds = (test_probs >= 0.5).astype(int)

metrics = {
    "Accuracy": accuracy_score(test_labels, test_preds),
    "Precision": precision_score(test_labels, test_preds),
    "Recall": recall_score(test_labels, test_preds),
    "F1_Score": f1_score(test_labels, test_preds),
    "AUC_ROC": roc_auc_score(test_labels, test_probs),
}
print("FT-Transformer performance (dropout = positive class):")
for k, v in metrics.items():
    print(f"  {k:10s}: {v:.4f}")

print()
print(classification_report(test_labels, test_preds, target_names=["Not Dropout", "Dropout"]))
```

```python
cm = confusion_matrix(test_labels, test_preds)
plt.figure(figsize=(4.5, 4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Pred: Not Dropout", "Pred: Dropout"],
            yticklabels=["True: Not Dropout", "True: Dropout"])
plt.title("Confusion Matrix — FT-Transformer")
plt.tight_layout(); plt.show()

fpr, tpr, _ = roc_curve(test_labels, test_probs)
plt.figure(figsize=(5, 4))
plt.plot(fpr, tpr, label=f"FT-Transformer (AUC = {metrics['AUC_ROC']:.3f})")
plt.plot([0, 1], [0, 1], "k--", alpha=0.4)
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curve"); plt.legend(); plt.tight_layout(); plt.show()
```

```python
bucket_results = []
for bucket in ["Early (1-3)", "Mid (4-5)", "Late (6-8)"]:
    mask = (bucket_test == bucket)
    if mask.sum() == 0:
        continue
    probs_b = test_probs[mask]
    labels_b = test_labels[mask]
    preds_b = (probs_b >= 0.5).astype(int)
    bucket_results.append({
        "Semester Window": bucket,
        "N": int(mask.sum()),
        "Accuracy": accuracy_score(labels_b, preds_b),
        "Precision": precision_score(labels_b, preds_b, zero_division=0),
        "Recall": recall_score(labels_b, preds_b, zero_division=0),
        "F1_Score": f1_score(labels_b, preds_b, zero_division=0),
        "AUC_ROC": roc_auc_score(labels_b, probs_b) if len(set(labels_b)) > 1 else np.nan,
    })

bucket_df = pd.DataFrame(bucket_results)
bucket_df
```

```python
import shap

feature_names = CATEGORICAL_COLS + NUMERIC_COLS

def predict_proba_fn(x):
    x = np.asarray(x)
    x_cat = torch.tensor(x[:, :n_cat], dtype=torch.long).to(DEVICE)
    x_num = torch.tensor(x[:, n_cat:], dtype=torch.float32).to(DEVICE)
    model.eval()
    with torch.no_grad():
        probs = torch.sigmoid(model(x_cat, x_num)).cpu().numpy()
    return probs

X_test_full = np.concatenate([X_test_cat, X_test_num], axis=1)

rng = np.random.default_rng(SEED)
background_idx = rng.choice(len(X_test_full), size=min(100, len(X_test_full)), replace=False)
explain_idx = rng.choice(len(X_test_full), size=min(100, len(X_test_full)), replace=False)

background = X_test_full[background_idx]
explain_set = X_test_full[explain_idx]

explainer = shap.KernelExplainer(predict_proba_fn, background)
shap_values = explainer.shap_values(explain_set, nsamples=100)
```

```python
# Global feature importance
shap.summary_plot(shap_values, explain_set, feature_names=feature_names, plot_type="bar",
                   show=False, max_display=20)
plt.tight_layout(); plt.show()
```

```python
# Local explanation for one individual student (advisor-facing)
student_row = 0
print(f"Predicted dropout probability: {predict_proba_fn(explain_set[student_row:student_row+1])[0]:.3f}")

shap.force_plot(
    explainer.expected_value, shap_values[student_row], explain_set[student_row],
    feature_names=feature_names, matplotlib=True, show=False,
)
plt.tight_layout(); plt.show()
```
